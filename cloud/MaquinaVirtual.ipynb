# ☁️ Máquina Virtual — Cloud Workstation

Workstation Linux temporária com perfil **MAX** automático. Usa todos os recursos que o runtime realmente disponibilizar, sem inventar núcleos, RAM ou GPU.

Arquitetura atual: `Colab → Xorg/Xvfb → GNOME/Flashback → x11vnc(localhost) → noVNC → HTTPS → app`.

> O runtime continua sujeito às regras e limites do provedor. Este notebook não contém anti-idle nem mecanismos para contornar encerramentos de sessão.


```python
#@title 1) Preparar projeto
import os, shutil, subprocess
ROOT = '/content/MaquinaVirtual'
if os.path.exists(ROOT):
    shutil.rmtree(ROOT)
subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/Johnatafgfdgf/M-quina-virtual.git', ROOT], check=True)
print('✅ Projeto carregado:', ROOT)
```

```python
#@title 2) Instalar desktop
import subprocess
subprocess.run(['bash', '/content/MaquinaVirtual/cloud/scripts/install.sh'], check=True)
print('✅ Dependências prontas.')
```

```python
#@title 3) Iniciar sessão — MAX PERFORMANCE
RESOLUTION = '1600x720' #@param {type:"string"}
import json, secrets, string, subprocess, html, shlex
from IPython.display import HTML, display
alphabet = string.ascii_letters + string.digits
PASSWORD = ''.join(secrets.choice(alphabet) for _ in range(8))
perf = '/content/MaquinaVirtual/cloud/scripts/performance.sh'
start = '/content/MaquinaVirtual/cloud/scripts/start.sh'
cmd = f"source {shlex.quote(perf)} && bash {shlex.quote(start)} {shlex.quote(RESOLUTION)} {shlex.quote(PASSWORD)}"
subprocess.run(['bash', '-lc', cmd], check=True)
with open('/tmp/maquina-virtual/session.json', encoding='utf-8') as f:
    session = json.load(f)
try:
    with open('/tmp/maquina-virtual/performance.json', encoding='utf-8') as f:
        perf_info = json.load(f)
except Exception:
    perf_info = {}
print('\n✅ SESSÃO PRONTA — PERFIL MAX')
print('CPU:', perf_info.get('cpu_threads', '?'), 'threads disponíveis')
print('RAM:', perf_info.get('ram_mb', '?'), 'MB')
print('GPU:', perf_info.get('gpu_name') or ('indisponível' if not perf_info.get('gpu_available') else 'NVIDIA'))
print('URL:', session['public_url'])
print('Senha temporária:', session['password'])
print('Resolução:', session['resolution'])
deep = html.escape(session['deep_link'], quote=True)
viewer = html.escape(session['viewer_url'], quote=True)
cpu = html.escape(str(perf_info.get('cpu_threads', '?')))
ram = html.escape(str(perf_info.get('ram_mb', '?')))
gpu = html.escape(str(perf_info.get('gpu_name') or 'CPU / llvmpipe'))
display(HTML(f'''<div style="font-family:system-ui;padding:18px;border-radius:18px;background:#111522;color:#fff;max-width:680px">
<div style="font-size:22px;font-weight:800;margin-bottom:8px">Máquina Virtual pronta · MAX</div>
<div style="color:#aab2cc;margin-bottom:10px">CPU {cpu} threads · RAM {ram} MB · GPU {gpu}</div>
<div style="color:#aab2cc;margin-bottom:16px">O perfil usa automaticamente o máximo de recursos realmente disponíveis no runtime.</div>
<a href="{deep}" style="display:inline-block;background:#7e5cff;color:white;padding:13px 18px;border-radius:12px;text-decoration:none;font-weight:700;margin-right:8px">ABRIR NO APP</a>
<a href="{viewer}" target="_blank" style="display:inline-block;background:#20283d;color:white;padding:13px 18px;border-radius:12px;text-decoration:none;font-weight:700">ABRIR NO NAVEGADOR</a>
</div>'''))
```
